Hücre 2: Görev 1 (G1) — Yalnızca NumPy ile Referans MLP Eğitimi
Ödev kılavuzundaki ipucuna uygun olarak gizli katmanda (H=128) ReLU aktivasyonu ve bias (b1) kullanılmış, G3'teki SNN dönüşümünü birebir eşleştirmek için çıkış katmanı bias'ı sıfırda sabitlenerek (b2 = 0) doğrusal bırakılmıştır.

```python
# G1: 64 -> H -> 10 yapısında yalnızca NumPy ile MLP
def relu(z):
  return np.maximum(0.0, z)


def softmax(z):
  z_kaydir = z - np.max(z, axis=1, keepdims=True)
  exp_z = np.exp(z_kaydir)
  return exp_z / np.sum(exp_z, axis=1, keepdims=True)


def mlp_ileri(X, W1, b1, W2, b2):
  z1 = X @ W1 + b1
  a1 = relu(z1)
  z2 = a1 @ W2 + b2  # Doğrusal çıkış katmanı (b2 = 0)
  olasilik = softmax(z2)
  return z1, a1, z2, olasilik


def mlp_egit(Xtr, ytr, H=128, epok=60, lr=0.1, batch_size=32, tohum=59):
  r = np.random.default_rng(tohum)
  N, D = Xtr.shape
  K = 10

  # He / Xavier ağırlık ilklendirmesi
  W1 = r.normal(0.0, np.sqrt(2.0 / D), size=(D, H))
  b1 = np.zeros(H, dtype=np.float64)
  W2 = r.normal(0.0, np.sqrt(1.0 / H), size=(H, K))
  b2 = np.zeros(
      K, dtype=np.float64
  )  # G3 dönüşümü için çıkış katmanı doğrusal (bias=0) bırakılır

  kayiplar = []

  for ep in range(epok):
    indisler = r.permutation(N)
    X_karisik = Xtr[indisler]
    y_karisik = ytr[indisler]

    epok_kaybi = 0.0
    for i in range(0, N, batch_size):
      Xb = X_karisik[i : i + batch_size]
      yb = y_karisik[i : i + batch_size]
      B = len(yb)

      # İleri yayılım
      z1, a1, z2, olasilik = mlp_ileri(Xb, W1, b1, W2, b2)

      # Çapraz entropi kaybı
      kayip = -np.mean(np.log(olasilik[np.arange(B), yb] + 1e-12))
      epok_kaybi += kayip * B

      # Geri yayılım
      dz2 = olasilik.copy()
      dz2[np.arange(B), yb] -= 1.0
      dz2 /= B

      dW2 = a1.T @ dz2
      da1 = dz2 @ W2.T
      dz1 = da1 * (z1 > 0.0)
      dW1 = Xb.T @ dz1
      db1 = np.sum(dz1, axis=0)

      # Mini-batch gradyan inişi güncellemesi
      W1 -= lr * dW1
      b1 -= lr * db1
      W2 -= lr * dW2

    kayiplar.append(epok_kaybi / N)

  return (W1, b1, W2, b2), kayiplar


# Modeli eğit ve test kümesinde değerlendir
agirliklar_mlp, mlp_kayiplar = mlp_egit(Xtr, ytr, H=128, epok=60, lr=0.1)
W1, b1, W2, b2 = agirliklar_mlp

_, _, _, y_olasilik_tr = mlp_ileri(Xtr, W1, b1, W2, b2)
_, _, _, y_olasilik_te = mlp_ileri(Xte, W1, b1, W2, b2)

y_tahmin_tr = np.argmax(y_olasilik_tr, axis=1)
y_tahmin_te = np.argmax(y_olasilik_te, axis=1)

acc_mlp_tr = np.mean(y_tahmin_tr == ytr) * 100
acc_mlp_te = np.mean(y_tahmin_te == yte) * 100

# ANN Enerji vekili hesabı
ann_mac, ann_pj = ann_enerji([64, 128, 10])

print(f"G1 Eğitim Doğruluğu : %{acc_mlp_tr:.2f}")
print(f"G1 Test Doğruluğu   : %{acc_mlp_te:.2f}")
print(
    f"ANN Örnek Başına MAC: {ann_mac} MAC | Enerji Vekili: {ann_pj:.1f} pJ"
    f" ({ann_pj/1000:.2f} nJ)"
)
```

Hücre 4: Görev 2 (G2) — Üç Spike Kodlayıcı ve Otomatik Test Doğrulaması
Bu hücrede poisson_kod, deterministik_kod (IF / Sigma-Delta) ve ttfs_kod (İlk Spike Zamanı) fonksiyonları (T, N, 64) çıktı boyutunda yazılır ve kodlayici_test_et ile doğrulanır.

```python
# G2: Spike Kodlayıcılar — Çıktı biçimi: (T, N, 64), değerler {0, 1}


def poisson_kod(x, T, tohum=42):
  """Poisson oran kodlayıcı: her adımda x olasılıkla spike üretir."""
  x = np.asarray(x, dtype=np.float64)
  r = np.random.default_rng(tohum)
  return (r.random((T,) + x.shape) < x).astype(np.int32)


def deterministik_kod(x, T):
  """Deterministik oran kodlayıcı (IF / Sigma-Delta): V += x, V >= 1 ise spike ve V -= 1."""
  x = np.asarray(x, dtype=np.float64)
  V = np.zeros_like(x, dtype=np.float64)
  spikelar = np.zeros((T,) + x.shape, dtype=np.int32)
  for t in range(T):
    V += x
    s = (V >= 1.0 - 1e-12).astype(np.int32)
    spikelar[t] = s
    V -= s * 1.0
  return spikelar


def ttfs_kod(x, T):
  """İlk spike zamanı (TTFS) kodlayıcı: güçlü piksel erken ateşler, her nöron en fazla 1 kez ateşler."""
  x = np.asarray(x, dtype=np.float64)
  spikelar = np.zeros((T,) + x.shape, dtype=np.int32)
  # Sadece pozitif yoğunluklu pikseller t = floor((1 - x) * (T - 1)) anında ateşler
  t_spike = np.clip(np.floor((1.0 - x) * (T - 1)).astype(int), 0, T - 1)
  aktif_maske = x > 0.0

  it = np.nditer(x, flags=["multi_index"])
  while not it.finished:
    idx = it.multi_index
    if aktif_maske[idx]:
      spikelar[(t_spike[idx],) + idx] = 1
    it.iternext()
  return spikelar


print("1) Poisson Kodlayıcı Testi:")
kodlayici_test_et(poisson_kod)

print("\n2) Deterministik (IF / Sigma-Delta) Kodlayıcı Testi:")
kodlayici_test_et(deterministik_kod)

print("\n3) TTFS (İlk Spike Zamanı) Kodlayıcı Testi:")
kodlayici_test_et(ttfs_kod)
```

```python
# 1. Bir test rakamı için üç kodun raster grafiği
ornek_idx = 0
x_ornek = Xte[ornek_idx : ornek_idx + 1]  # Biçim: (1, 64)
etiket_ornek = yte[ornek_idx]
T_raster = 50

s_poisson = poisson_kod(x_ornek, T_raster)[
    :, 0, :
]  # (T, 64)
s_det = deterministik_kod(x_ornek, T_raster)[:, 0, :]
s_ttfs = ttfs_kod(x_ornek, T_raster)[:, 0, :]

fig, axes = plt.subplots(1, 4, figsize=(16, 4), gridspec_kw={"width_ratios": [1, 2, 2, 2]})

axes[0].imshow(x_ornek.reshape(8, 8), cmap="gray_r")
axes[0].set_title(f"Test Örneği #{ornek_idx}\n(Rakam: {etiket_ornek})")
axes[0].axis("off")

kod_verileri = [
    ("Poisson Oran Kodu", s_poisson, "tab:blue"),
    ("Deterministik (IF) Kodu", s_det, "tab:orange"),
    ("TTFS (İlk Spike) Kodu", s_ttfs, "tab:green"),
]

for ax, (baslik, s_mat, renk) in zip(axes[1:], kod_verileri):
  t_ind, noron_ind = np.where(s_mat == 1)
  ax.scatter(t_ind, noron_ind, s=12, color=renk, marker="|")
  ax.set_title(f"{baslik}\nToplam Spike: {int(s_mat.sum())}")
  ax.set_xlabel("Zaman Adımı (t)")
  ax.set_ylabel("Piksel / Nöron İndisi (0-63)")
  ax.set_xlim(-1, T_raster)
  ax.set_ylim(-2, 65)
  ax.grid(True, alpha=0.25)

plt.tight_layout()
plt.show()

# 2. T in {10, 20, 50, 100, 200} için yeniden oluşturma hatası (Log-Log)
T_listesi = [10, 20, 50, 100, 200]
hata_poisson = []
hata_det = []

for T_deger in T_listesi:
  # 3 farklı tohumun ortalamasıyla Poisson yeniden oluşturma hatası
  p_hatalar = [
      np.abs(poisson_kod(Xte, T_deger, tohum=s).mean(axis=0) - Xte).mean()
      for s in [10, 20, 30]
  ]
  hata_poisson.append(np.mean(p_hatalar))

  # Deterministik kod yeniden oluşturma hatası
  d_hata = np.abs(deterministik_kod(Xte, T_deger).mean(axis=0) - Xte).mean()
  hata_det.append(d_hata)

# Log-log uzayda doğrusal regresyon ile deneysel eğim hesabı
egim_poisson = np.polyfit(np.log10(T_listesi), np.log10(hata_poisson), 1)[0]
egim_det = np.polyfit(np.log10(T_listesi), np.log10(hata_det), 1)[0]

print(f"Poisson Oran Kodu Deneysel Log-Log Eğimi       : {egim_poisson:.3f} (Teorik 1/√T beklentisi: -0.500)")
print(f"Deterministik Oran Kodu Deneysel Log-Log Eğimi : {egim_det:.3f} (Teorik 1/T beklentisi : -1.000)")

# Log-Log Grafik
plt.figure(figsize=(8, 5))
plt.loglog(
    T_listesi,
    hata_poisson,
    "o-",
    lw=2,
    label=f"Poisson Kodu (Deneysel eğim = {egim_poisson:.2f})",
)
plt.loglog(
    T_listesi,
    hata_det,
    "s-",
    lw=2,
    label=f"Deterministik IF Kodu (Deneysel eğim = {egim_det:.2f})",
)

# Teorik referans doğruları
T_arr = np.array(T_listesi, dtype=float)
plt.loglog(
    T_arr,
    hata_poisson[0] * (T_arr / T_arr[0]) ** (-0.5),
    "--",
    color="gray",
    alpha=0.7,
    label=r"Teorik $1/\sqrt{T}$ (eğim = -0.5)",
)
plt.loglog(
    T_arr,
    hata_det[0] * (T_arr / T_arr[0]) ** (-1.0),
    ":",
    color="black",
    alpha=0.7,
    label=r"Teorik $1/T$ (eğim = -1.0)",
)

plt.xticks(T_listesi, labels=[str(t) for t in T_listesi])
plt.xlabel("Zaman Penceresi Uzunluğu (T) [log ölçek]")
plt.ylabel("Ortalama Yeniden Oluşturma Hatası |oran - x| [log ölçek]")
plt.title("G2: Oran Kodlarının Yeniden Oluşturma Hatası (Log-Log)")
plt.legend()
plt.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()
```

Hücre 6: Görev 3 (G3) — Çıkarma ile Sıfırlamalı IF Katmanı ve Otomatik TestiBu hücrede çıkarma ile sıfırlama (reset by subtraction: $V_t = V_{t-1} + z_t - s_t \cdot \theta$) kullanan if_katmani fonksiyonu yazılır ve if_test_et ile ateşleme oranının ReLU (clip(z, 0, 1)) fonksiyonuna yakınsadığı doğrulanır.

```python
# G3: Çıkarma ile sıfırlama (reset by subtraction) kullanan IF nöron katmanı
def if_katmani(z, esik=1.0):
  """z: (T, ...) boyutunda zaman serisi girdisi.

  Çıktı: z ile aynı boyutta {0, 1} spike dizisi (reset by subtraction).
  """
  z = np.asarray(z, dtype=np.float64)
  T = z.shape[0]
  V = np.zeros(z.shape[1:], dtype=np.float64)
  spikelar = np.zeros(z.shape, dtype=np.int32)

  for t in range(T):
    V += z[t]
    s = (V >= esik - 1e-12).astype(np.int32)
    spikelar[t] = s
    V -= s * esik  # Çıkarma ile sıfırlama (reset by subtraction)

  return spikelar


print("IF Katmanı Otomatik Doğrulama Testi:")
if_test_et(if_katmani)
```

Hücre 7: Veri Tabanlı Normalizasyon ($\lambda$ = 99,9. Yüzdelik) ve SNN KoşturucuÖdev kılavuzundaki uyarıya uygun olarak normalizasyon yüzdelikleri ($\lambda_1, \lambda_2$) yalnızca eğitim verisi (Xtr) üzerinden hesaplanır. Ayrıca enerji hesabında hem giriş katmanı (l=0) hem de gizli katman (l=1) spike'ları eksiksiz sayılır.

```python
# Yalnızca eğitim verisi (Xtr) üzerinden 99.9. yüzdelik (lambda) hesabı
z1_tr, a1_tr, z2_tr, _ = mlp_ileri(Xtr, W1, b1, W2, b2)

lambda1 = np.percentile(a1_tr, 99.9)
lambda2 = np.percentile(np.maximum(0.0, z2_tr), 99.9)

# Ağırlık ve bias'ların veri tabanlı normalizasyonla ölçeklenmesi (Rueckauer vd., 2017)
W1_norm = W1 / lambda1
b1_norm = b1 / lambda1
W2_norm = W2 * (lambda1 / lambda2)
b2_norm = b2 / lambda2

agirliklar_norm = (W1_norm, b1_norm, W2_norm, b2_norm)

print(f"Eğitim verisi 99.9. yüzdelik aktivasyon ölçekleri:")
print(f"  • Gizli katman (λ1) : {lambda1:.4f}")
print(f"  • Çıkış katmanı (λ2): {lambda2:.4f}")


def snn_calistir(X, y, T, agirliklar, kodlayici=poisson_kod, tohum=42):
  """Dönüştürülmüş IF SNN ağını T adım çalıştırır.

  Dönüş: doğruluk (%), [ort_giris_spike, ort_gizli_spike], synop, enerji_pJ
  """
  W1_k, b1_k, W2_k, b2_k = agirliklar

  # 1. Giriş spike kodlaması: (T, N, 64)
  if kodlayici is poisson_kod:
    s_giris = kodlayici(X, T, tohum=tohum)
  else:
    s_giris = kodlayici(X, T)

  # 2. Gizli katman akımı ve IF nöronları: (T, N, 128)
  z1_snn = s_giris @ W1_k + b1_k
  s_gizli = if_katmani(z1_snn, esik=1.0)

  # 3. Çıkış katmanı akımı ve IF nöronları: (T, N, 10)
  z2_snn = s_gizli @ W2_k + b2_k
  s_cikis = if_katmani(z2_snn, esik=1.0)

  # Karar: çıkıştaki toplam spike sayısı en yüksek olan sınıf
  toplam_cikis_spike = s_cikis.sum(axis=0)  # (N, 10)
  y_tahmin = np.argmax(toplam_cikis_spike, axis=1)
  dogruluk = np.mean(y_tahmin == y) * 100.0

  # Örnek başına ortalama toplam spike sayıları (Giriş katmanı l=0 ve Gizli katman l=1)
  ort_giris_spike = s_giris.sum(axis=(0, 2)).mean()
  ort_gizli_spike = s_gizli.sum(axis=(0, 2)).mean()
  spike_sayilari = [ort_giris_spike, ort_gizli_spike]

  # snn_enerji ile SynOp ve pJ hesabı ([64, 128, 10])
  synop, enerji_pj = snn_enerji(spike_sayilari, [64, 128, 10])

  return dogruluk, spike_sayilari, synop, enerji_pj
```

Hücre 10: Görev 4 (G4) — Vekil Türev ve BPTT ile LIF SNN Eğitimi ($\text{Eğim} \in \{1, 10, 100\}$ ve Gerçek Türev Ablasyonu)Ödev kılavuzunda verilen vekil(V, egim) hızlı sigmoid türevi kullanılarak $T=25$ adım ve $\beta=0.9$ sızıntı katsayısı ile 64 → 128 → 10 LIF ağı zamanda açılımlı geri yayılımla (BPTT) eğitilir. Ablasyon deneyinde gerçek türev (0) kullanıldığında ne olduğu hem yalnızca gizli katman türevi sıfırlandığında (ders kurgusu: ~%75–82) hem de tüm katmanlarda sıfırlandığında (~%10) ayrı ayrı gösterilir.

```python
# G4: Vekil Türevle Doğrudan Eğitilen LIF Ağı (BPTT, T=25, beta=0.9)


def vekil(V, egim):
  """Hızlı sigmoid vekil türevi: dS/dV ≈ 1 / (1 + egim * |V - 1|)^2"""
  return 1.0 / (1.0 + egim * np.abs(V - 1.0)) ** 2


def gercek_turev(V, egim=0):
  """Gerçek Heaviside türevi: her yerde 0 (Ablasyon testi için)"""
  return np.zeros_like(V)


def lif_ileri(s_giris, W1, b1, W2, beta=0.9, esik=1.0):
  """LIF SNN ileri yayılımı (T, B, 64) -> gizli ve çıkış spike/potansiyel kayıtları."""
  T, B, _ = s_giris.shape
  H = W1.shape[1]
  K = W2.shape[1]

  V1 = np.zeros((B, H), dtype=np.float64)
  V2 = np.zeros((B, K), dtype=np.float64)

  V1_tilde_kayit = np.zeros((T, B, H), dtype=np.float64)
  s1_kayit = np.zeros((T, B, H), dtype=np.float64)
  V2_tilde_kayit = np.zeros((T, B, K), dtype=np.float64)
  s2_kayit = np.zeros((T, B, K), dtype=np.float64)

  for t in range(T):
    # Gizli katman LIF dinamiği
    I1 = s_giris[t] @ W1 + b1
    V1_tilde = beta * V1 + I1
    s1 = (V1_tilde >= esik).astype(np.float64)
    V1 = V1_tilde - s1 * esik  # Çıkarma ile sıfırlama

    # Çıkış katmanı LIF dinamiği
    I2 = s1 @ W2
    V2_tilde = beta * V2 + I2
    s2 = (V2_tilde >= esik).astype(np.float64)
    V2 = V2_tilde - s2 * esik

    V1_tilde_kayit[t] = V1_tilde
    s1_kayit[t] = s1
    V2_tilde_kayit[t] = V2_tilde
    s2_kayit[t] = s2

  return V1_tilde_kayit, s1_kayit, V2_tilde_kayit, s2_kayit


def lif_degerlendir(X, y, agirliklar, T=25, beta=0.9, tohum=42):
  """LIF modelini test eder; doğruluk, spike sayıları, SynOp ve pJ döndürür."""
  W1_l, b1_l, W2_l = agirliklar
  s_giris = poisson_kod(X, T, tohum=tohum).astype(np.float64)
  _, s1_kayit, _, s2_kayit = lif_ileri(s_giris, W1_l, b1_l, W2_l, beta=beta)

  # Çıkışta toplam spike sayısına göre karar
  cikis_sayim = s2_kayit.sum(axis=0)  # (N, 10)
  y_tahmin = np.argmax(cikis_sayim, axis=1)
  acc = np.mean(y_tahmin == y) * 100.0

  ort_giris_sp = s_giris.sum(axis=(0, 2)).mean()
  ort_gizli_sp = s1_kayit.sum(axis=(0, 2)).mean()
  synop, pj = snn_enerji([ort_giris_sp, ort_gizli_sp], [64, 128, 10])
  return acc, [ort_giris_sp, ort_gizli_sp], synop, pj


def lif_bptt_egit(
    Xtr,
    ytr,
    T=25,
    H=128,
    beta=0.9,
    egim=10,
    epok=40,
    lr=0.004,
    batch_size=32,
    ablasyon_modu=None,
    tohum=59,
):
  """Zamanda Açılımlı Geri Yayılım (BPTT) ve Adam güncellemeli LIF SNN eğitimi.

  ablasyon_modu: None -> normal vekil türev 'gizli_sifir' -> gizli katmanda
  gerçek türev (0), çıkışta hata sinyali açık (ders kurgusu) 'tam_sifir' -> tüm
  katmanlarda gerçek türev (0)
  """
  r = np.random.default_rng(tohum)
  N, D = Xtr.shape
  K = 10

  # Ağırlık ilklendirmesi
  W1 = r.normal(0.0, np.sqrt(2.0 / D), size=(D, H))
  b1 = np.zeros(H, dtype=np.float64)
  W2 = r.normal(0.0, np.sqrt(2.0 / H), size=(H, K))

  # Adam moment değişkenleri
  mW1, vW1 = np.zeros_like(W1), np.zeros_like(W1)
  mb1, vb1 = np.zeros_like(b1), np.zeros_like(b1)
  mW2, vW2 = np.zeros_like(W2), np.zeros_like(W2)
  beta1_adam, beta2_adam, eps = 0.9, 0.999, 1e-8
  adim = 0

  kayip_gecmisi = []

  for ep in range(epok):
    indisler = r.permutation(N)
    X_shuf = Xtr[indisler]
    y_shuf = ytr[indisler]

    # Her epokta Poisson spike üretimi
    s_giris_tum = poisson_kod(X_shuf, T, tohum=tohum + ep).astype(np.float64)
    epok_kayip = 0.0

    for i in range(0, N, batch_size):
      sb = s_giris_tum[:, i : i + batch_size, :]  # (T, B, 64)
      yb = y_shuf[i : i + batch_size]
      B = len(yb)

      # İleri yayılım
      V1_tilde, s1, V2_tilde, s2 = lif_ileri(sb, W1, b1, W2, beta=beta)

      # Çıkış toplam spike sayısı üzerinden Softmax + Çapraz Entropi
      cikis_oran = s2.sum(axis=0) / (T / 5.0)  # Ölçeklenmiş spike toplamı
      olasilik = softmax(cikis_oran)
      kayip = -np.mean(np.log(olasilik[np.arange(B), yb] + 1e-12))
      epok_kayip += kayip * B

      # Çıkış hatası: dL / d(s2_t)
      d_cikis = olasilik.copy()
      d_cikis[np.arange(B), yb] -= 1.0
      d_cikis /= B * (T / 5.0)

      # BPTT: t = T-1 ... 0
      dW1 = np.zeros_like(W1)
      db1 = np.zeros_like(b1)
      dW2 = np.zeros_like(W2)

      delta1_next = np.zeros((B, H), dtype=np.float64)
      delta2_next = np.zeros((B, K), dtype=np.float64)

      for t in range(T - 1, -1, -1):
        # Çıkış katmanı türevi
        if ablasyon_modu == "tam_sifir":
          psi2 = gercek_turev(V2_tilde[t])
        elif ablasyon_modu == "gizli_sifir":
          psi2 = np.ones_like(V2_tilde[t])  # Çıkışta doğrudan hata geçişi
        else:
          psi2 = vekil(V2_tilde[t], egim)

        delta2 = d_cikis * psi2 + beta * delta2_next
        dW2 += s1[t].T @ delta2

        # Gizli katman türevi
        gs1 = delta2 @ W2.T
        if ablasyon_modu in ("gizli_sifir", "tam_sifir"):
          psi1 = gercek_turev(V1_tilde[t])  # Gerçek türev = 0!
        else:
          psi1 = vekil(V1_tilde[t], egim)

        delta1 = gs1 * psi1 + beta * delta1_next
        dW1 += sb[t].T @ delta1
        db1 += np.sum(delta1, axis=0)

        delta2_next = delta2
        delta1_next = delta1

      # Adam güncellemesi
      adim += 1
      for param, grad, m, v_mom in [
          (W1, dW1, mW1, vW1),
          (b1, db1, mb1, vb1),
          (W2, dW2, mW2, vW2),
      ]:
        m[:] = beta1_adam * m + (1.0 - beta1_adam) * grad
        v_mom[:] = beta2_adam * v_mom + (1.0 - beta2_adam) * (grad**2)
        m_hat = m / (1.0 - beta1_adam**adim)
        v_hat = v_mom / (1.0 - beta2_adam**adim)
        param -= lr * m_hat / (np.sqrt(v_hat) + eps)

    kayip_gecmisi.append(epok_kayip / N)

  return (W1, b1, W2), kayip_gecmisi


# 1) Üç farklı vekil türev eğimi (1, 10, 100) ile eğitim ve 3 tohumlu test
egimler = [1, 10, 100]
g4_sonuclar = {}

print("G4: Vekil Türev Eğimi Karşılaştırması (T=25, β=0.9):")
print("-" * 78)
for eg in egimler:
  ag_lif, kayip_lif = lif_bptt_egit(
      Xtr, ytr, T=25, H=128, beta=0.9, egim=eg, epok=40, lr=0.004
  )
  accs, synops, pjs, sp_list = [], [], [], []
  for s in [101, 202, 303]:
    acc_s, sp_s, syn_s, pj_s = lif_degerlendir(
        Xte, yte, ag_lif, T=25, beta=0.9, tohum=s
    )
    accs.append(acc_s)
    synops.append(syn_s)
    pjs.append(pj_s)
    sp_list.append(sp_s)

  g4_sonuclar[eg] = {
      "agirliklar": ag_lif,
      "kayiplar": kayip_lif,
      "acc_mean": np.mean(accs),
      "acc_std": np.std(accs),
      "synop": np.mean(synops),
      "pj": np.mean(pjs),
      "spikelar": np.mean(sp_list, axis=0),
  }
  print(
      f"  • Vekil Eğim = {eg:>3} | Test Doğruluğu:"
      f" %{g4_sonuclar[eg]['acc_mean']:.2f} ±{g4_sonuclar[eg]['acc_std']:.2f} |"
      f" SynOp: {g4_sonuclar[eg]['synop']:.1f} | Enerji:"
      f" {g4_sonuclar[eg]['pj']:.1f} pJ"
  )

# 2) Ablasyon: Gerçek Türev (Her yerde 0) Deneyi
ag_abl_gizli, kayip_abl_gizli = lif_bptt_egit(
    Xtr, ytr, T=25, H=128, beta=0.9, epok=40, lr=0.004, ablasyon_modu="gizli_sifir"
)
acc_abl_gizli = np.mean([
    lif_degerlendir(Xte, yte, ag_abl_gizli, T=25, beta=0.9, tohum=s)[0]
    for s in [101, 202, 303]
])

ag_abl_tam, kayip_abl_tam = lif_bptt_egit(
    Xtr, ytr, T=25, H=128, beta=0.9, epok=40, lr=0.004, ablasyon_modu="tam_sifir"
)
acc_abl_tam = np.mean([
    lif_degerlendir(Xte, yte, ag_abl_tam, T=25, beta=0.9, tohum=s)[0]
    for s in [101, 202, 303]
])

print("-" * 78)
print(
    "  • Ablasyon (Gizli katmanda gerçek türev=0, W1 donuk / sadece W2 öğrenir)"
    f" : %{acc_abl_gizli:.2f} (Dersteki %76.5 referansının karşılığı)"
)
print(
    "  • Ablasyon (Tüm katmanlarda gerçek türev=0, W1 ve W2 tamamen donuk)     "
    f" : %{acc_abl_tam:.2f} (Rastgele tahmin seviyesi)"
)
```

Hücre 12: Bonus Görev (+10 Puan) — Beyincik (Granül–Purkinje) Yaklaşımı ile Yerel ÖğrenmeÖdevin Bonus bölümündeki (a) Beyincik yaklaşımı uygulanmıştır: Gizli katmandaki LIF nöronları (granül hücreleri) sabit ve rastgele ağırlıklara (W1_sabit) sahiptir; geriye hiçbir hata yayılmaz. Yalnızca çıkış ağırlıkları (W2), tırmanıcı lif hata sinyali $(y - \hat{y})$ ve granül hücrelerinin ateşleme oranı $r^{(1)}$ kullanılarak tamamen yerel delta kuralı ($\Delta W_2 = \eta \, r^{(1)\top} (y - \hat{y})$) ile güncellenir. 

```python
# BONUS (+10 Puan): Beyincik Yaklaşımı — Sabit Rastgele Granül Katmanı + Yerel Delta Kuralı


def beyincik_lif_ozellik(X, W1_sabit, b1_sabit, T=25, beta=0.9, tohum=42):
  """Sabit rastgele ağırlıklı LIF granül katmanının ateşleme oranlarını (paralel lif çıktısı) hesaplar."""
  s_giris = poisson_kod(X, T, tohum=tohum).astype(np.float64)
  T_adim, B, _ = s_giris.shape
  H = W1_sabit.shape[1]

  V1 = np.zeros((B, H), dtype=np.float64)
  toplam_s1 = np.zeros((B, H), dtype=np.float64)

  for t in range(T_adim):
    V1_tilde = beta * V1 + (s_giris[t] @ W1_sabit + b1_sabit)
    s1 = (V1_tilde >= 1.0).astype(np.float64)
    V1 = V1_tilde - s1
    toplam_s1 += s1

  giris_sp_ort = s_giris.sum(axis=(0, 2)).mean()
  gizli_sp_ort = toplam_s1.sum(axis=1).mean()
  return (
      toplam_s1 / T_adim,
      [giris_sp_ort, gizli_sp_ort],
  )  # (B, H) granül ateşleme oranı


def beyincik_egit(Xtr, ytr, H=128, T=25, beta=0.9, epok=60, lr=0.15, batch_size=32, tohum=59):
  r = np.random.default_rng(tohum)
  N, D = Xtr.shape
  K = 10

  # 1. Sabit rastgele granül katmanı ağırlıkları (Hiç güncellenmez!)
  W1_sabit = r.normal(0.0, np.sqrt(2.0 / D), size=(D, H))
  b1_sabit = r.uniform(-0.2, 0.2, size=H)

  # 2. Eğitilebilir Purkinje / çıkış ağırlıkları
  W2_yerel = np.zeros((H, K), dtype=np.float64)
  b2_yerel = np.zeros(K, dtype=np.float64)

  kayiplar_bonus = []

  for ep in range(epok):
    # Granül hücrelerinin paralel lif ateşleme oranları r1: (N, H)
    r1_tr, _ = beyincik_lif_ozellik(
        Xtr, W1_sabit, b1_sabit, T=T, beta=beta, tohum=tohum + ep
    )

    indisler = r.permutation(N)
    r1_shuf = r1_tr[indisler]
    y_shuf = ytr[indisler]
    ep_kayip = 0.0

    for i in range(0, N, batch_size):
      rb = r1_shuf[i : i + batch_size]
      yb = y_shuf[i : i + batch_size]
      B = len(yb)

      # Çıkış tahmini ve tırmanıcı lif hata sinyali: e = y_gercek - y_tahmin
      olasilik = softmax(rb @ W2_yerel + b2_yerel)
      ep_kayip += -np.sum(np.log(olasilik[np.arange(B), yb] + 1e-12))

      y_onehot = np.zeros((B, K), dtype=np.float64)
      y_onehot[np.arange(B), yb] = 1.0
      hata_tirmanici_lif = (y_onehot - olasilik) / B  # Yerel hedef hata sinyali

      # Yerel Delta Kuralı: ΔW2 = lr * (r1^T @ hata) -> Geri yayılım YOK!
      W2_yerel += lr * (rb.T @ hata_tirmanici_lif)
      b2_yerel += lr * np.sum(hata_tirmanici_lif, axis=0)

    kayiplar_bonus.append(ep_kayip / N)

  return (W1_sabit, b1_sabit, W2_yerel, b2_yerel), kayiplar_bonus


# Aynı gizli boyut (H=128) ve genişletilmiş granül katmanı (H=512, biyolojik beyincik mimarisi) ile test
for H_granul in [128, 512]:
  ag_beyincik, kayip_beyincik = beyincik_egit(Xtr, ytr, H=H_granul, T=25, epok=60, lr=0.2)
  W1_b, b1_b, W2_b, b2_b = ag_beyincik

  accs_b, syn_b, pj_b = [], [], []
  for s in [101, 202, 303]:
    r1_te, sp_b = beyincik_lif_ozellik(Xte, W1_b, b1_b, T=25, tohum=s)
    tahmin_b = np.argmax(r1_te @ W2_b + b2_b, axis=1)
    accs_b.append(np.mean(tahmin_b == yte) * 100.0)
    sy, pj_val = snn_enerji(sp_b, [64, H_granul, 10])
    syn_b.append(sy)
    pj_b.append(pj_val)

  if H_granul == 128:
    ag_beyincik_128 = ag_beyincik
    acc_beyincik_128 = np.mean(accs_b)
    std_beyincik_128 = np.std(accs_b)
    syn_beyincik_128 = np.mean(syn_b)
    pj_beyincik_128 = np.mean(pj_b)

  print(
      f"Bonus Beyincik Modeli (H={H_granul:>3} Granül Hücresi) | "
      f"Test Doğruluğu: %{np.mean(accs_b):.2f} ±{np.std(accs_b):.2f} | "
      f"SynOp: {np.mean(syn_b):.1f} | Enerji: {np.mean(pj_b):.1f} pJ"
  )
```
